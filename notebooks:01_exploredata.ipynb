```python
import pandas as pd
import numpy as np
import yfinance as yf
```

```python
data = yf.download("SPY GLD TSM BNP.PA 0700.HK XOM NESN.SW EURUSD=X HKDUSD=X CHFUSD=X",start="2016-01-01", end="2026-11-06")
```

```python
prices = data.Close
print(prices.isna().sum())
print(prices.isna().any(axis=1))
```

```python
prices_usd = prices.copy()
prices_usd['BNP.PA'] = prices_usd['BNP.PA'] * prices_usd['EURUSD=X']
prices_usd['0700.HK'] = prices_usd['0700.HK'] * prices_usd['HKDUSD=X']
prices_usd['NESN.SW'] = prices_usd['NESN.SW'] * prices_usd['CHFUSD=X']
prices_usd.tail()
```

```python
returns = prices_usd.dropna().pct_change().dropna()
returns.head()
```
